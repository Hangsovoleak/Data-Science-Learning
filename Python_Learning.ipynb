```python
# import math library for using with `floor` function
import math

# Rounding Numbers
price = 35.548793
print(round(price))
# round(number, ndigits) : rounds the number to the specified number of decimal places
print(round(price, 2))
print(math.floor(price))

print(math.ceil(price))

# math.trunc(x) - cuts off the decimal part and keeps the whole number (no rounding)
print(math.trunc(price))
print(int(price))

# Notes: int() vs trunc()
# if you're not using math already, just use int() it's simple and built-in
# if you've already imported math, use trunc() it makes your intent clearer
```

```python
# How to Sort?
l = ['c', 'a', 'b']
# l.sort()
l.sort(reverse=True)
print(l)

matrix = [
    ['d', 'e', 'f'],
    ['g', 'h', 'i'],
    ['a', 'a', 'c']
]

# matrix.sort()
print('Original matrix:', matrix)

new_matrix = sorted(matrix)
print('Sorted matrix:', new_matrix)

new_matrix_a = sorted(matrix, reverse = True)
print('Sorted matrix:', new_matrix_a)
```

```python
# Interator
indexs = ['a', 'b', 'c']
new_index = []
for l in indexs:
  new_index.append(l.upper())
  print(new_index)

print()
print(list(enumerate(indexs, start = 1)))
print()
for index, value in enumerate(indexs):
  print(index, value)

# reversed
print()
for i in reversed(indexs):
  print(i)

# zip
print()
numbers = [1, 2, 3]
for l, n in zip(indexs, numbers):
  print(l, n)

# map
print()
print(map(str.upper, indexs))
print(list(map(str.upper, indexs)))

# map with str.strip
names = [' Porty ', 'Lyhorng ', ' Davin']
for n in map(str.strip, names):
  print(n)

print()
mixes = ['a', '', 'b', None, 'c', False]
print(list(filter(None, mixes)))
print(list(filter(bool, mixes)))

items = ['sql', '123', 'python', '42']
# print(list(filter(str.isalpha, items)))
print()

for i in filter(str.isalpha, items):
  print(i)

# Lambda
print()
multiple = lambda x: x*2
print(multiple(2))

add = lambda x, y: x + y
print(add(3, 6))
print()

check = lambda i: i in "python"
print(check('n'))

# Prices are stored as messay strings and need cleaning to floats
prices = ['$12.50', '$9.99', '$100.00']
lambda p: float(p.replace('$', ''))

p = '$12.50'
print(float(p.replace('$', '')))
print(list(map(lambda p: float(p.replace('$', '')), prices)))


price = [120, 30, 300, 80]

print(list(filter(lambda p: p >= 100, price)))
print()

students = [
    ['Porty', 85],
    ['Lyhorng', 90],
    ['Davin', 60]
]

print(list(filter(lambda row: row[1] > 70, students)))

print(students[1][1] > 70)
```
